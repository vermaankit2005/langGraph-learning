```python
from dotenv import load_dotenv
from langchain.agents import create_agent
from langchain_core.messages import HumanMessage
from langchain_groq import ChatGroq
from langchain_tavily import TavilySearch
from langchain_mcp_adapters.client import MultiServerMCPClient


load_dotenv()

SERVERS = {
    "manim-server": {
      "transport": "stdio",
      "command": "C:\\Users\\verma\\AppData\\Local\\Programs\\Python\\Python313\\python.exe",
      "args": [
        "D:\\mcp-servers\\manim-mcp-server\\src\\manim_server.py"
      ],
      "env": {
        "MANIM_EXECUTABLE": "C:\\Users\\verma\\AppData\\Local\\Programs\\Python\\Python313\\Scripts\\manim.exe"
      }
    }
}

mcp_client = MultiServerMCPClient(SERVERS)

tools = await mcp_client.get_tools()

print("Available tools:")
for tool in tools:
    print(f"- {tool.name}: {tool.description}")


search_tool = TavilySearch(
    max_results=1,
    topic="general",
    include_images=False,
    search_depth="basic"
)
llm = ChatGroq(model="openai/gpt-oss-120b", temperature=0.4, reasoning_format="hidden")

agent = create_agent(model=llm, tools=[search_tool])
```

```python
streaming_iterator = agent.stream(
    input={"messages": [HumanMessage(content="tell a horror story")]}, stream_mode="messages")

last_node = None

for message, metadata in streaming_iterator:
    if not message.content:
        continue

    node = metadata.get("langgraph_node")

    if node != last_node:
        last_node = node
        print(f"\n\n--- {node} ---")

    print(message.content, end="", flush=True)

```
